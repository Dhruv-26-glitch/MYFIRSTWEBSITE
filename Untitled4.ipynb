```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler


df = pd.DataFrame({
    "income": [15000,18000, 20000,22000,25000,30000,100000]
})

scaler=MinMaxScaler()

df["Scaled_income"]=scaler.fit_transform(df[["income"]])

print(df)
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler

df = pd.DataFrame({
    "Age": [ 0,30,40],
    "Salary": [20000,50000,80000]
})
print(df)

scaler = StandardScaler()
scaled_data  = scaler.fit_transform(df)
df_scaled = pd.DataFrame(scaled_data, columns=df.columns)
print(df_scaled)
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, MinMaxScaler, LabelEncoder

df = pd.DataFrame({

    "Name" : ["A", "B", "C", "D" ,"E"],
    "Gender" :["Female", "Female", "Female", "Male", "Male"],
    "Math" : [85,79,92,70,60],
    "science" : [90,83,34,47,93],
    "English" : [60,70,65,69,45],
    "Pass" : [1,1,1,0,0]
})

#1. Encode Gender

le= LabelEncoder()
df["Gender_encoded"] = le.fit_transform(df["Gender"])
print(df)

#2. Featured Scaling (Math,Science, English )

scaler = MinMaxScaler()
df[["Math_scaled", "Science_scaled", "English_scaled"]] = scaler.fit_transform(df[["Math", "science", "English"]])
print(df)
```

```python
from sklearn.feature_selection import SelectKBest, f_classif

# 3. Feature selection ( Select top 2 features )
# Features to consider (scaled + encoded)
X = df[["Gender_encoded", "Math_scaled", "Science_scaled", "English_scaled"]]
Y = df["Pass"]

# Select top 2 features
selector = SelectKBest (score_func=f_classif, k=2)
X_new = selector.fit_transform(X,Y)

selected_fetures = X.columns[selector.get_support()]
print(selected_fetures)
print("Selected Features : ", list(selected_fetures ))
print("Transformed Data: \n", X_new)
```
